```python
import math

def e_approx(n):
    s = 0
    fact = 1
    for i in range(n+1):
        if i > 0:
            fact *= i
        s += 1 / fact
    return s

true_e = math.e

for n in [5, 10, 15, 20]:
    approx = e_approx(n)
    abs_error = abs(true_e - approx)
    rel_error = abs_error / abs(true_e)
    
    print(f"n={n}")
    print(f"e* = {approx}")
    print(f"błąd bezwzględny = {abs_error}")
    print(f"błąd względny = {rel_error}")
    print()
```

Zadanie 6

```python
def newton(f, f_prim, x_0, tol=1e-10, max_iter=100):
    x = x_0
    for _ in range(max_iter):
        x_new = x - f(x)/f_prim(x)
        if abs(x_new - x) < tol:
            return x_new
        x = x_new
    return x

def f(x):
    return x**2 - 2

def f_prim(x):
    return 2*x

wynik = newton(f, f_prim, 1.0)
print("Pierwiastek z 2 ≈", wynik)
```

```python
import math

def G(p, pi_approx):
    return (1/p) * (10**p * (1 + p*pi_approx * 10**(-p)) - 10**p)

for p in range(1, 20):
    val = G(p, math.pi)
    print(p, val)
```
