Zadanie 2

```python
import numpy as np

M = np.array([
    [np.sqrt(2), 1, -np.sqrt(2)],[0, 1, 1],[-np.sqrt(2), np.sqrt(2), 1]
])

M_inv = np.linalg.inv(M)

A_T = M.T

det_M = np.linalg.det(M)

print(f"Macierz M: {M}")


print(f"\nMacierz odwrotna M^-1: {M_inv}")

print(f"\nMacierz transponowana: {A_T}")

print(f"\nWyznacznik macierzy M: {det_M}")
```

Zadanie 3

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**3 - 5*x + 7


x1 = np.linspace(-1, 1, 100)
y1 = f(x1)


x2 = np.linspace(-10, 20, 200)
y2 = f(x2)


plt.figure()
plt.plot(x1, y1)
plt.title("f(x) = x^3 - 5x + 7 (zakres -1 do 1)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()


plt.figure()
plt.plot(x2, y2)
plt.title("f(x) = x^3 - 5x + 7 (zakres -10 do 20)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()

plt.show()

```

Zadanie 5

```python
import numpy as np
from scipy.linalg import qr

A = np.random.rand(10, 10)

Q, R = qr(A)


Q_T = Q.T
Q_inv = np.linalg.inv(Q)


print("Q:\n", Q)
print("\nR:\n", R)

print("\nQ^T:\n", Q_T)
print("\nQ^-1:\n", Q_inv)

print("\nCzy Q^T = ", np.allclose(Q_T, Q_inv))
```
